# Figure 1a: MRVA study area

Draws the MRVA study-area map with topography, rivers, and AEM flight lines.


```python
from pathlib import Path
import os
import subprocess
import sys
from datetime import datetime

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "data").exists():
    REPO_ROOT = next(p for p in REPO_ROOT.parents if (p / "data").exists())

SCRIPT_PATH = REPO_ROOT / "tools" / "plot_fig1a_mrva_study_system_gmt.py"
OUTPUT_DIR = REPO_ROOT / "outputs" / "figures" / "Fig1a_MRVA_study_system_GMT"
OUTPUT_PREFIX = OUTPUT_DIR / "Fig1a_MRVA_study_system_GMT"
outputs = {
    "png": OUTPUT_PREFIX.with_suffix(".png"),
    "pdf": OUTPUT_PREFIX.with_suffix(".pdf"),
    "work_dir": OUTPUT_DIR,
}

FIG1A_PARAMS = {
    "FIG1A_REGION": "-94/-87/30.5/38.5",
    "FIG1A_CACHE_SUBDIR": "figure_cache",
    "FIG1A_MAP_WIDTH_CM": "12.4",
    "FIG1A_DPI": "600",
    "FIG1A_DEM_SIZE": "1200,1700",

    "FIG1A_MAP_FRAME_TYPE": "fancy",
    "FIG1A_MAP_FRAME_WIDTH": "4p",
    "FIG1A_MAP_FRAME_PEN": "1.0p,black",
    "FIG1A_MAP_FRAME_PERCENT": "100",
    "FIG1A_MAP_TICK_LENGTH_PRIMARY": "4p/2p",

    "FIG1A_FONT_SCALE": "1.35",
    "FIG1A_FONT_ADD_PT": "3.0",

    "FIG1A_STATE_LABEL_FONT_SCALE": "1.0",
    "FIG1A_STATE_LABEL_FONT_ADD_PT": "3",
    "FIG1A_STATE_LABEL_COLOR": "#0f0e0e",

    "FIG1A_SHADE_INTENSITY": "0.9",
    "FIG1A_TERRAIN_TRANSPARENCY": "10",

    "FIG1A_MRVA_FILL_CYAN": "none",
    "FIG1A_MRVA_FILL_YELLOW": "none",
    "FIG1A_BOUNDARY_PEN": "0.85p,white",
    "FIG1A_STATE_BOUNDARY_PEN": "0.50p,#4a4a4a",

    "FIG1A_AEM_SEGMENT_STRIDE": "1",
    "FIG1A_AEM_VERTEX_STRIDE": "1",
    "FIG1A_AEM_LINE_PEN": "0.8p, #111111",
    "FIG1A_DRAW_WELLS": "0",
    "FIG1A_DRAW_STATE_LABELS": "0",
    "FIG1A_DRAW_LEGEND": "0",

    "FIG1A_TRIBUTARY_ORDER1_PEN": "0.52p,#75b9d4@50",
    "FIG1A_TRIBUTARY_ORDER2_PEN": "0.26p,#9fd3e7@66",
    "FIG1A_MISSISSIPPI_HALO_PEN": "none",
    "FIG1A_MISSISSIPPI_PEN": "0.1p,#045a8d",

    "FIG1A_LEGEND_WIDTH_CM": "4.90",
    "FIG1A_LEGEND_OFFSET": "0.85c/1.95c",
    "FIG1A_SCALEBAR_OFFSET": "1.25c/0.88c",
    "FIG1A_NORTH_ARROW_SIZE_CM": "1.25",
    "FIG1A_NORTH_ARROW_OFFSET": "0.35c/0.32c",
}

print("Fig. 1a adjustable parameters:")
for key, value in FIG1A_PARAMS.items():
    print(f"  {key} = {value}")

for key in ["png", "pdf"]:
    if outputs[key].exists():
        outputs[key].unlink()

env = os.environ.copy()
env.update(FIG1A_PARAMS)
result = subprocess.run(
    [sys.executable, str(SCRIPT_PATH)],
    cwd=str(REPO_ROOT),
    env=env,
    text=True,
    capture_output=True,
    check=True,
)
print(result.stdout)
if result.stderr.strip():
    print(result.stderr)

for key in ["png", "pdf"]:
    path = outputs[key]
    stamp = datetime.fromtimestamp(path.stat().st_mtime).strftime("%Y-%m-%d %H:%M:%S")
    print(f"{key.upper()}: {path} | modified: {stamp} | size: {path.stat().st_size:,} bytes")

legend_path = outputs["work_dir"] / "legend.txt"
if legend_path.exists():
    print("\nCurrent legend file:")
    print(legend_path.read_text(encoding="utf-8"))
else:
    print("\nLegend drawing is disabled; no legend.txt was written.")
outputs
```
