# Figure 1a: location map

Shows the position of the MRVA within the conterminous United States.


```python
from pathlib import Path
import os
import subprocess
import sys
from datetime import datetime

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "data").exists():
    REPO_ROOT = next(p for p in REPO_ROOT.parents if (p / "data").exists())

SCRIPT_PATH = REPO_ROOT / "tools" / "plot_fig1a_position_gmt.py"
OUTPUT_DIR = REPO_ROOT / "outputs" / "figures" / "Fig1a_Position_GMT"
OUTPUT_PREFIX = OUTPUT_DIR / "Fig1a_Position_GMT"
outputs = {
    "png": OUTPUT_PREFIX.with_suffix(".png"),
    "pdf": OUTPUT_PREFIX.with_suffix(".pdf"),
    "work_dir": OUTPUT_DIR,
}

FIG1A_POS_PARAMS = {
    "FIG1A_POS_REGION": "-125/-66/24/50",
    "FIG1A_POS_STUDY_REGION": "-94/-87/30.5/38.5",
    "FIG1A_POS_MAP_WIDTH_CM": "7.2",
    "FIG1A_POS_DPI": "600",
    "FIG1A_POS_PROJECTION": "B-96/37/29.5/45.5/7.20c",

    "FIG1A_POS_LAND_FILL": "#fbfbfb",
    "FIG1A_POS_WATER_FILL": "white",
    "FIG1A_POS_USA_OUTLINE_PEN": "0.45p,#cfcfcf",
    "FIG1A_POS_STATE_PEN": "0.28p,#d8d8d8",

    "FIG1A_POS_BOX_FILL": "#efb190@36",
    "FIG1A_POS_BOX_PEN": "0.55p,#ba6a52",
    "FIG1A_POS_MRVA_FILL": "#c97154@30",
    "FIG1A_POS_MRVA_PEN": "0.65p,#3b2018",
}

env = os.environ.copy()
env.update(FIG1A_POS_PARAMS)

print("Fig. 1a position locator parameters:")
for key, value in FIG1A_POS_PARAMS.items():
    print(f"  {key} = {value}")

subprocess.run([sys.executable, str(SCRIPT_PATH)], cwd=REPO_ROOT, env=env, check=True)

for key in ["png", "pdf"]:
    path = outputs[key]
    if path.exists():
        mtime = datetime.fromtimestamp(path.stat().st_mtime).strftime("%Y-%m-%d %H:%M:%S")
        print(f"{key.upper()}: {path} | modified: {mtime} | size: {path.stat().st_size:,} bytes")
    else:
        print(f"Missing {key.upper()}: {path}")

try:
    from IPython.display import Image, display
    display(Image(filename=str(outputs["png"])))
except Exception:
    pass
```
